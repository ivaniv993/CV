```python
import cv2
import dlib
import numpy as np
from matplotlib import pyplot as plt
print(cv2.__version__)
```

```python
video_path = 'slow_traffic_small.mp4'



# init_box = cv2.selectROI("Tracking", frame, fromCenter=False, showCrosshair=True)
init_box = [225, 129, 50, 37];
print('cordinates : ', init_box)

def trackerTest(tracker, video_path, init_box, lost_threshold):
    cap = cv2.VideoCapture(video_path)
    suc, frame = cap.read()
    frame_idx = 0
    tracker_cords = []
    tracker.init(frame, init_box)

    # if 5 frames in row marked as losted than interrupt
    break_in_case_of_lost = lost_threshold
    while True:
        success, frame = cap.read()
        if not success or break_in_case_of_lost == 0:
            break

        # every 10 frames
        if (frame_idx % 10 == 0):
            ok, box = tracker.update(frame)
            if ok:
                tracker_cords.append(box)
                x, y, w, h = [int(v) for v in box]
                cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)
                cv2.putText(frame, "Tracking", (10, 30),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0,255,0), 2)

                break_in_case_of_lost = lost_threshold
            else:
                cv2.putText(frame, "Lost", (10, 30),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0,0,255), 2)
                break_in_case_of_lost -= 1
                
            cv2.imshow("Tracking", frame)
    
        frame_idx += 1
    
        if cv2.waitKey(30) & 0xFF == ord('q'):
            break
    cap.release()
    cv2.destroyAllWindows()
    return tracker_cords

# KCF
trackerKCF = cv2.legacy.TrackerKCF_create()
cordinatesKCF = trackerTest(trackerKCF, video_path, init_box, 10)
print('cordinatesKCF : ', cordinatesKCF)

# CSRT
cap = cv2.VideoCapture(video_path)
trackerCSRT = cv2.legacy.TrackerCSRT_create()
cordinatesCSRT = trackerTest(trackerCSRT, video_path, init_box, 5)
print('cordinatesCSRT : ', cordinatesCSRT)

# Do you see any differences? If so, what are they? 
#        KCF tracker is not able track car when it scaled up.
# Does one tracker perform better than the other? In what way?
#        CSRT - works better in this case. It tracked car along the road. 
#        But actually it will continue track car after it disappear from screen (after few frames)
```
