```python
import cv2
import dlib
import numpy as np
from matplotlib import pyplot as plt
print(cv2.__version__)
```

```python
video_path = 'slow_traffic_small.mp4'



# init_box = cv2.selectROI("Tracking", frame, fromCenter=False, showCrosshair=True)
init_box = [225, 129, 50, 37];
print('cordinates : ', init_box)

def trackerTest(tracker, video_path, init_box):
    cap = cv2.VideoCapture(video_path)
    suc, frame = cap.read()
    frame_idx = 0
    tracker_cords = []
    tracker.init(frame, init_box)
    while True:
        success, frame = cap.read()
        if not success:
            break
    
        if (frame_idx % 10 == 0):
            ok, box = tracker.update(frame)
            if ok:
                tracker_cords.append(box)
                x, y, w, h = [int(v) for v in box]
                cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)
                cv2.putText(frame, "Tracking", (10, 30),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0,255,0), 2)
            else:
                cv2.putText(frame, "Lost", (10, 30),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0,0,255), 2)
            cv2.imshow("Tracking", frame)
    
        frame_idx += 1
    
        if cv2.waitKey(30) & 0xFF == ord('q'):
            break
    cap.release()
    cv2.destroyAllWindows()
    return tracker_cords

# KCF
trackerKCF = cv2.legacy.TrackerKCF_create()
cordinatesKCF = trackerTest(trackerKCF, cap, init_box)
print('cordinatesKCF : ', cordinatesKCF)

# CSRT
cap = cv2.VideoCapture(video_path)
trackerCSRT = cv2.legacy.TrackerCSRT_create()
cordinatesCSRT = trackerTest(trackerCSRT, cap, init_box)
print('cordinatesCSRT : ', cordinatesCSRT)
```
